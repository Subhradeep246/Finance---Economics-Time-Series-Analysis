```python
import pandas as pd
import numpy as np
```

```python
# Getting a feel for the data
df_raw = pd.read_csv('finance_economics_dataset.csv')
df_raw.head()
```

```python
# Checking null values
df_raw.isnull().sum()
```

```python
# Getting number of Stock Indices being tracked
stock_indices = df_raw["Stock Index"].unique()
stock_indices
```

```python
# Separating the data by Stock Index
df_DJ = df_raw[df_raw["Stock Index"] == "Dow Jones"]
df_SnP = df_raw[df_raw["Stock Index"] == "S&P 500"]
df_NAS = df_raw[df_raw["Stock Index"] == "NASDAQ"]
# df_NAS
```

```python
# Getting the date range of the data
min_date = df_raw.Date.min()
max_date = df_raw.Date.max()
expected_range = pd.date_range(start = min_date, end = max_date, freq='D')
expected_range
```

```python
# Getting the number of missing dates for each index
missing_days_DJ = expected_range.difference(pd.DatetimeIndex(df_DJ["Date"]))
missing_days_SnP = expected_range.difference(pd.DatetimeIndex(df_SnP["Date"]))
missing_days_NAS = expected_range.difference(pd.DatetimeIndex(df_NAS["Date"]))
```

```python
# Converting Date string data to Datetime objects:
df_DJ['Date'] = pd.to_datetime(df_DJ['Date'])
df_SnP['Date'] = pd.to_datetime(df_SnP['Date'])
df_NAS['Date'] = pd.to_datetime(df_NAS['Date'])

# Adding missing days to each dataframe:
column_names = df_raw.columns

df_missing_DJ = pd.DataFrame(columns=column_names)
df_missing_DJ["Date"] = missing_days_DJ
df_DJ_con = pd.concat([df_DJ, df_missing_DJ], ignore_index=True)

df_missing_SnP = pd.DataFrame(columns=column_names)
df_missing_SnP["Date"] = missing_days_SnP
df_SnP_con = pd.concat([df_SnP, df_missing_SnP], ignore_index=True)

df_missing_NAS = pd.DataFrame(columns=column_names)
df_missing_NAS["Date"] = missing_days_NAS
df_NAS_con = pd.concat([df_NAS, df_missing_NAS])

# Sorting Dataframes and redoing indices:
df_DJ_con.sort_values(by='Date', ascending=True, inplace=True)
df_DJ_con.reset_index(drop=True, inplace=True)

df_SnP_con.sort_values(by='Date', ascending=True, inplace=True)
df_SnP_con.reset_index(drop=True, inplace=True)

df_NAS_con.sort_values(by='Date', ascending=True, inplace=True)
df_NAS_con.reset_index(drop=True, inplace=True)
```

```python
# Method to fill dataframes
def fill_na_vals(df, method):
    ans = pd.DataFrame(columns=df.columns)
    match method:
        case "F":
            ans = df.fillna(method="ffill", axis=0)
        case "B":
            ans = df.fillna(method="bfill", axis=0)
    return ans
# Filling both ways to ensure there are no null values remaining
fill_method = "F"
df_DJ_filled = fill_na_vals(df_DJ_con, fill_method)
df_SnP_filled = fill_na_vals(df_SnP_con, fill_method)
df_NAS_filled = fill_na_vals(df_NAS_con, fill_method)

fill_method = "B"
df_DJ_filled = fill_na_vals(df_DJ_filled, fill_method)
df_SnP_filled = fill_na_vals(df_SnP_filled, fill_method)
df_NAS_filled = fill_na_vals(df_NAS_filled, fill_method)
```

```python
# Checking that there are no null values remaining
df_SnP_filled.isnull().sum()
# df_DJ_filled.isnull().sum()
# df_NAS_filled.isnull().sum()
```

```python
# Combining the cleaned dataframes together and then sorting by date
df_cleaned = pd.concat([df_DJ_filled, df_SnP_filled, df_NAS_filled], ignore_index=True)
df_cleaned.sort_values(by='Date', ascending=True, inplace=True)
df_cleaned.reset_index(drop=True, inplace=True)
df_cleaned.to_csv(r"C:\Users\Justin Camargo\OneDrive\Documents\School Docs - NYU\2025 Fall\Data Science Bootcamp\Finance---Economics-Time-Series-Analysis\Cleaned Data\dataset_cleaned.csv", index=False)

df_DJ_filled.to_csv(r"C:\Users\Justin Camargo\OneDrive\Documents\School Docs - NYU\2025 Fall\Data Science Bootcamp\Finance---Economics-Time-Series-Analysis\Cleaned Data\Dow_Jones_data_cleaned.csv", index=False)
df_SnP_filled.to_csv(r"C:\Users\Justin Camargo\OneDrive\Documents\School Docs - NYU\2025 Fall\Data Science Bootcamp\Finance---Economics-Time-Series-Analysis\Cleaned Data\S_and_P_500_data_cleaned.csv", index=False)
df_NAS_filled.to_csv(r"C:\Users\Justin Camargo\OneDrive\Documents\School Docs - NYU\2025 Fall\Data Science Bootcamp\Finance---Economics-Time-Series-Analysis\Cleaned Data\NASDAQ_data_cleaned.csv", index=False)
```
